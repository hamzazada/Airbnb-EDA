```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
data = pd.read_csv('airbnb.csv', encoding_errors='ignore')
```

```python
data.isnull().sum()

#dropping All missing values
data.dropna(inplace=True)

data.isnull().sum()
```

```python
data.shape
```

```python
data.duplicated().sum()

# data[data.duplicated()]
data.drop_duplicates(inplace=True)

data.duplicated().sum()

data[data.duplicated()]
```

```python
data.dtypes

data['id'] = data['id'].astype('object')

data['host_id'] = data['host_id'].astype('object')
data.dtypes
```

```python
#identifying outliers in price column
df = data[data['price'] <1500]
sns.boxplot(data = df, x = 'price')
```

```python
#price distribution
plt.figure(figsize=(8,5))
sns.histplot(data=df, x='price' , bins=50)
plt.title('Price Distribution')
plt.ylabel('frequency')
plt.show()
```

```python
df.dtypes
```

```python
#price distribution
plt.figure(figsize=(6,3))
sns.histplot(data=df, x='availability_365')
plt.title('availability_365 Distribution')
plt.ylabel('frequency')
plt.show()
```

```python
data.dtypes
```

```python
#average price per bed 
df.groupby('neighbourhood_group')['price per beds'].mean()
```

```python
df.groupby('neighbourhood_group')['price'].mean()
```

```python
#'price per beds'
df['price per beds'] = df['price']/df['beds']
df.head()
```

** Bi variable analysis**
one variable dependancy in another variable

```python
df.columns
```

```python
# price dependancy on neighbourhood
sns.barplot(data=df, x='neighbourhood_group', y='price', hue='room_type')
```

```python
#number of reviews and price relation
plt.figure(figsize=(8,5))
plt.title('Locality and Review Dependancy')
sns.scatterplot(data=df, x= 'number_of_reviews', y='price', hue='neighbourhood_group')
plt.show()
```

```python
df.dtypes
```

```python
sns.pairplot(data=df, vars=['price', 'minimum_nights', 'number_of_reviews', 'availability_365'], hue='room_type')
```

```python
#Geographical Distribution of Airbnb Listing 
plt.figure(figsize=(10,7))
sns.scatterplot(data=df , x='longitude', y='latitude',hue='room_type')
plt.title('Geographical Distribution of Airbnb Listing ')
plt.show()
```

```python
#heat map - correlation of variable

corr = df[[
    'latitude', 'longitude', 'price', 'minimum_nights', 'number_of_reviews',
    'reviews_per_month', 'availability_365', 'beds'
]].corr()
corr
plt.Figure(figsize=(8,8))
sns.heatmap(data=corr, annot=True)
```
